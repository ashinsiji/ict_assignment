1. Write a Python function to fi nd the Max of three numbers

```python
def find_max(a, b, c):
    return max(a, b, c)
print(find_max(10, 25, 15))
```

2. Write a Python function to sum all the numbers in a list.

```python
def sum_list(numbers):
    total = 0
    for x in numbers:
        total += x
    return total
sample_list = [18, 12, 30, 10, 7]
print(sum_list(sample_list))
```

3. Python program to insert a number to any position in a list (Get the number to be inserted and position from the user)

```python
my_list = [1, 2, 3, 4, 5]
num = int(input("Enter the number to insert: "))
pos = int(input("Enter the position (index): "))
my_list.insert(pos, num)
print("Updated list:", my_list)
```

4. Write a program to rename a key city to a location in the following dictionary.

```python
sample_dict = {"name": "Kelly", "age": 25, "salary": 8000, "city": "New york"}
sample_dict['location'] = sample_dict.pop('city')
print(sample_dict)
```
