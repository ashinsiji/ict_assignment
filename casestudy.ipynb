📌 PART 7 — Age Analysis

Q14. Age Distribution

Plot histogram of age of medal winners.


What is the most common winning age range?




```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.histplot(medal_winners['Age'].dropna(), bins=20, kde=True)
plt.title('Distribution of Age Among Medal Winners')
plt.xlabel('Age')
plt.ylabel('Number of Medal Winners')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
age_counts = medal_winners['Age'].value_counts(bins=range(10, 60, 5)).sort_index()
print("Frequency of ages (in 5-year bins):")
display(age_counts)
most_common_age_range = age_counts.idxmax()
print(f"\nThe most common winning age range is: {most_common_age_range}")
```

Q15. Average Age by Medal Type
Find average age for Gold, Silver, and Bronze medal winners.


```python
average_age_by_medal = medal_winners.groupby('Medal')['Age'].mean().reset_index()
average_age_by_medal.columns = ['Medal', 'Average_Age']

print('Average Age for Each Medal Type:')
display(average_age_by_medal)

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))
sns.barplot(x='Medal', y='Average_Age', data=average_age_by_medal, palette='coolwarm')
plt.title('Average Age of Medal Winners by Medal Type')
plt.xlabel('Medal Type')
plt.ylabel('Average Age')
plt.tight_layout()
plt.show()
```

📌 PART 8 — Correlation & Numerical Insight


Q16. Correlation Matrix

Create correlation heatmap for numerical columns (Age, Height, Weight).


Which two variables are most correlated?




```python
import matplotlib.pyplot as plt
import seaborn as sns
correlation_data = medal_winners[['Age', 'Height', 'Weight']].dropna()
correlation_matrix = correlation_data.corr()
print('Correlation Matrix:')
display(correlation_matrix)
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap of Age, Height, and Weight for Medal Winners')
plt.tight_layout()
plt.show()
import numpy as np
mask = correlation_matrix.where(np.triu(np.ones(correlation_matrix.shape), k=1).astype(bool))
most_correlated_pair = mask.stack().abs().idxmax()
correlation_value = mask.stack().loc[most_correlated_pair]
print(f"The two most correlated variables are '{most_correlated_pair[0]}' and '{most_correlated_pair[1]}' with a correlation of {correlation_value:.2f}.")
```

Q17. Country Performance

Choose one country:

Total medals


Gold medals


```python
usa_medals = medal_winners[medal_winners['NOC'] == 'USA']
display(usa_medals.head())
print(f"Shape of usa_medals DataFrame: {usa_medals.shape}")
total_medals_usa = usa_medals.shape[0]
print(f"Total medals won by USA: {total_medals_usa}")
usa_gold_medals = usa_medals[usa_medals['Medal'] == 'Gold']
total_gold_medals_usa = usa_gold_medals.shape[0]
print(f"Total gold medals won by USA: {total_gold_medals_usa}")
usa_medals_per_sport = usa_medals['Sport'].value_counts().reset_index()
usa_medals_per_sport.columns = ['Sport', 'Medal_Count']

highest_medal_sport_usa = usa_medals_per_sport.sort_values(by='Medal_Count', ascending=False).iloc[0]

print(f"The sport in which USA has won the highest number of medals is: {highest_medal_sport_usa['Sport']} with {highest_medal_sport_usa['Medal_Count']} medals.")
usa_medal_trend_over_years = usa_medals.groupby('Year')['Medal'].count().reset_index()
usa_medal_trend_over_years.columns = ['Year', 'Medal_Count']

print('USA Medal Trend Over Years:')
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 6))
sns.lineplot(x='Year', y='Medal_Count', data=usa_medal_trend_over_years)
plt.title('USA Medal Trend Over Years')
plt.xlabel('Year')
plt.ylabel('Medal Count')
plt.grid(True)
plt.tight_layout()
plt.show()
```

Q20. Medal Ratio

For top 5 countries:

Calculate Gold / Total medal ratio.


```python
medal_ratio_df = pd.merge(medal_counts_by_country, gold_medal_counts_by_country, on='NOC', how='left')
medal_ratio_df['Gold_Medal_Count'] = medal_ratio_df['Gold_Medal_Count'].fillna(0)
medal_ratio_df['Gold_Ratio'] = medal_ratio_df['Gold_Medal_Count'] / medal_ratio_df['Medal_Count']
top_5_countries_ratio = medal_ratio_df.sort_values(by='Medal_Count', ascending=False).head(5)
print('Gold to Total Medal Ratio for Top 5 Countries:')
display(top_5_countries_ratio[['NOC', 'Gold_Ratio']])
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(10, 6))
sns.barplot(x='NOC', y='Gold_Ratio', data=top_5_countries_ratio, palette='magma')
plt.title('Gold to Total Medal Ratio for Top 5 Countries')
plt.xlabel('Country (NOC)')
plt.ylabel('Gold Medal Ratio')
plt.ylim(0, 1)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
